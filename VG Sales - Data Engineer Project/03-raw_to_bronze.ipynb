#### `COPY INTO` 

_Adicionando novos dados na tabela `fact_vg_sales_main`

```python
%sql
COPY INTO vg_sales.01_bronze.fact_vg_sales_main
FROM (
  SELECT 
    *,
    _metadata.file_path AS source_file, 
    current_timestamp() AS ingested_at
  FROM '/Volumes/vg_sales/01_bronze/volume_bronze' 
)
FILEFORMAT = CSV
FORMAT_OPTIONS (
  'header' = 'true', 
  'inferSchema' = 'false', 
  'sep' = ',', 
  'multiLine' = 'true'
)
COPY_OPTIONS (
  'mergeSchema' = 'true', 
  'force' = 'false'
);
```

```python
%sql

OPTIMIZE vg_sales.01_bronze.fact_vg_sales_main
ZORDER BY (ingested_at); 
```
